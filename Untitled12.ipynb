```python


import numpy as np
import cv2
import matplotlib.pyplot as plt
from google.colab import files


def load_image_from_upload():
    uploaded = files.upload()
    if not uploaded:
        raise ValueError("No image uploaded. Please upload a binary or grayscale object image.")
    img_path = next(iter(uploaded.keys()))
    img = cv2.imread(img_path, cv2.IMREAD_GRAYSCALE)
    if img is None:
        raise ValueError("Failed to read image. Make sure it's a valid image file.")
    return img

def binarize(img):
    _, bw = cv2.threshold(img, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)
    if np.sum(bw == 255) < np.sum(bw == 0):
        bw = 255 - bw
    return bw

def extract_main_contour(bw):
    cnts, _ = cv2.findContours(bw, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_NONE)
    if not cnts:
        raise Exception("No contours found. Please upload a valid binary image.")
    cnt = max(cnts, key=cv2.contourArea)
    return cnt[:, 0, :]

def resample_contour(cnt, N=512):
    pts = cnt.astype(np.float32)
    d = np.sqrt(np.sum(np.diff(pts, axis=0, append=pts[:1]) ** 2, axis=1))
    s = np.concatenate([[0], np.cumsum(d)])[:-1]
    total = s[-1] + d[-1]
    t = np.linspace(0, total, N, endpoint=False) % total
    x = np.interp(t, s, pts[:, 0], period=total)
    y = np.interp(t, s, pts[:, 1], period=total)
    return np.stack([x, y], axis=1)

def fourier_descriptors(cnt, normalize=True):
    z = cnt[:, 0] + 1j * cnt[:, 1]
    z -= np.mean(z)
    F = np.fft.fft(z)
    if not normalize:
        return F
    k1 = 1 if abs(F[1]) > 1e-12 else np.where(abs(F) > 1e-12)[0][1]
    scale = abs(F[k1])
    F /= scale
    theta = np.angle(F[k1])
    return F * np.exp(-1j * theta)

def reconstruct_contour(F, K=10):
    N = len(F)
    G = np.zeros_like(F)
    G[0] = F[0]
    for k in range(1, K + 1):
        G[k] = F[k]
        G[-k] = F[-k]
    z = np.fft.ifft(G)
    return np.stack([np.real(z), np.imag(z)], axis=1)


print("Please upload a grayscale or binary image (white object on black background):")
img_gray = load_image_from_upload()
bw = binarize(img_gray)
cnt = extract_main_contour(bw)
cntN = resample_contour(cnt, N=512)
F = fourier_descriptors(cntN, normalize=True)

Ks = [5, 10, 20, 40]
recons = [(K, reconstruct_contour(F, K)) for K in Ks]

plt.figure(figsize=(14, 10))

plt.subplot(2, 4, 1)
plt.imshow(img_gray, cmap='gray')
plt.title("Original")
plt.axis("off")

plt.subplot(2, 4, 2)
plt.imshow(bw, cmap='gray')
plt.title("Binary Image + Contour")
plt.plot(cntN[:, 0], cntN[:, 1], '.r', markersize=1)
plt.axis("off")

plt.subplot(2, 4, 3)
plt.plot(np.abs(np.fft.fftshift(F)))
plt.title("Magnitude of Fourier Descriptors")
plt.grid(True)

for i, (K, xy) in enumerate(recons, start=4):
    plt.subplot(2, 4, i)
    plt.plot(cntN[:, 0], cntN[:, 1], label='Original', linewidth=1)
    plt.plot(xy[:, 0], xy[:, 1], label=f'Recon K={K}', linewidth=2)
    plt.gca().invert_yaxis()
    plt.axis('equal')
    plt.title(f"Reconstruction with K={K}")
    plt.legend()

plt.tight_layout()
plt.show()
```
